### Experiment-2: Basic Text Preprocessing


Experiment 2.1: Write a Python program to perform basic text preprocessing on a given data. The program should:

- Convert text to lowercase. (Also display how many uppercase?)
- Remove punctuation. (Also display how many different types of punctuations?)
- Remove numbers. (Also display how many numbers?)
- Remove extra whitespace. (How much extra whitespace is present?)


```python
import re
import string
from collections import Counter

text = """Natural Language Processing (NLP) is a branch of Artificial Intelligence.  It helps computers understand and process human language.

In 2026, NLP is used in many applications such as chatbots, machine translation, sentiment analysis, and text summarization!!!

Text preprocessing is an important step in NLP.   It includes several operations such as cleaning text, removing punctuation, and removing unnecessary numbers.

For example, a chatbot may process 1000+ customer messages every day. These messages can contain CAPITAL letters, numbers like 12345, punctuation marks!!!, and   extra spaces.

The main goal of text preprocessing is to convert raw text into a clean and consistent format. This makes the text easier to analyze and use for further Natural Language Processing tasks."""
```

- Convert text to lowercase. (Also display how many uppercase?)


```python
uppercase_count = sum(1 for ch in text if ch.isupper())
print("1. Number of uppercase letters:", uppercase_count)
cleaned_text = text.lower()
```

- Remove punctuation. (Also display how many different types of punctuations?)


```python
punctuation_chars = [ch for ch in text if ch in string.punctuation]
punctuation_count = len(punctuation_chars)

punctuation_types = Counter(punctuation_chars)

print("\n2. Total punctuation marks:", punctuation_count)
print("Different types of punctuation:", len(punctuation_types))
print("Punctuation and their frequencies:")

for punctuation, count in punctuation_types.items():
    print(f"   '{punctuation}' : {count}")

cleaned_text = cleaned_text.translate(
    str.maketrans("", "", string.punctuation)
)
```

- Remove numbers. (Also display how many numbers?)


```python
digit_count = sum(1 for ch in text if ch.isdigit())

number_tokens = re.findall(r'\d+', text)

print("\n3. Total digits:", digit_count)
print("Number tokens:", len(number_tokens))
print("Numbers found:", number_tokens)

cleaned_text = re.sub(r'\d+', '', cleaned_text)
```

- Remove extra whitespace. (How much extra whitespace is present?)


```python
extra_whitespace_matches = re.findall(r'\s{2,}', cleaned_text)

extra_whitespace_count = sum(
    len(match) - 1 for match in extra_whitespace_matches
)

print("\n4. Extra whitespace characters:", extra_whitespace_count)

cleaned_text = re.sub(r'\s+', ' ', cleaned_text).strip()
```

- Display the cleaned text.


```python
print("\n5. CLEANED TEXT:")
print(cleaned_text)
```

Experiment 2.2: Write a Python program to perform word tokenization and sentence tokenization on a given text. Display the resulting words and sentences separately.
(Three separate code: using NLTK, using spaCy and without using NLTK/spacy)


- Using NLTK


```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')

text = """Natural Language Processing (NLP) is a fascinating field of Artificial Intelligence. It enables computers to understand human language. NLP is used in many real-world applications, such as chatbots, machine translation, sentiment analysis, and text summarization.

Text tokenization is one of the first steps in NLP. Sentence tokenization divides a paragraph into individual sentences. Word tokenization divides each sentence into individual words and tokens.

For example, a chatbot receives thousands of messages every day. The system processes these messages and identifies useful information. This information can then be used to provide better responses to users."""

sentences = nltk.sent_tokenize(text)

words = nltk.word_tokenize(text)

print("========== SENTENCE TOKENS ==========\n")

for i, sentence in enumerate(sentences, 1):
    print(f"{i}. {sentence}")

print("\n========== WORD TOKENS ==========\n")

for i, word in enumerate(words, 1):
    print(f"{i}. {word}")
```

- Using spaCy


```python
import spacy

nlp = spacy.blank("en")

nlp.add_pipe("sentencizer")

text = """Natural Language Processing (NLP) is a fascinating field of Artificial Intelligence. It enables computers to understand human language. NLP is used in many real-world applications, such as chatbots, machine translation, sentiment analysis, and text summarization.

Text tokenization is one of the first steps in NLP. Sentence tokenization divides a paragraph into individual sentences. Word tokenization divides each sentence into individual words and tokens.

For example, a chatbot receives thousands of messages every day. The system processes these messages and identifies useful information. This information can then be used to provide better responses to users."""

doc = nlp(text)

print("========== SENTENCE TOKENS ==========\n")

for i, sentence in enumerate(doc.sents, 1):
    print(f"{i}. {sentence.text}")

print("\n========== WORD TOKENS ==========\n")

for i, token in enumerate(doc, 1):
    print(f"{i}. {token.text}")
```

- Without NLTK or SpaCy


```python
import re

text = """Natural Language Processing (NLP) is a fascinating field of Artificial Intelligence. It enables computers to understand human language. NLP is used in many real-world applications, such as chatbots, machine translation, sentiment analysis, and text summarization.

Text tokenization is one of the first steps in NLP. Sentence tokenization divides a paragraph into individual sentences. Word tokenization divides each sentence into individual words and tokens.

For example, a chatbot receives thousands of messages every day. The system processes these messages and identifies useful information. This information can then be used to provide better responses to users."""

sentences = re.split(r'(?<=[.!?])\s+', text.strip())

print("========== SENTENCE TOKENS ==========\n")

for i, sentence in enumerate(sentences, 1):
    print(f"{i}. {sentence}")


words = re.findall(r'\w+|[^\w\s]', text)

print("\n========== WORD TOKENS ==========\n")

for i, word in enumerate(words, 1):
    print(f"{i}. {word}")
```

Experiment 2.3: Write a Python program to tokenize a text and remove commonly occurring stop words such as is, the, a, an, in, of, and, to, etc. Display the original and filtered tokens. (Also, display Stop Words Separately).


```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

text = """Natural Language Processing is an important field of Artificial Intelligence.
It is used in many applications such as chatbots, machine translation, sentiment analysis, and text summarization.

The main purpose of NLP is to enable computers to understand and process human language.
A computer can analyze a large amount of text in a short period of time.
The removal of stop words is one of the common steps in text preprocessing.

Stop words are commonly occurring words such as the, is, a, an, in, of, and, to, for, on, with, from, by, this, that, and it.
Removing these words can reduce the size of the text and improve the efficiency of some NLP applications."""

tokens = word_tokenize(text)

print("========== ORIGINAL TOKENS ==========\n")
print(tokens)

stop_words = set(stopwords.words('english'))
found_stop_words = [
    token for token in tokens
    if token.lower() in stop_words
]

print("\n========== STOP WORDS ==========\n")
print(found_stop_words)

filtered_tokens = [
    token for token in tokens
    if token.lower() not in stop_words
]

print("\n========== FILTERED TOKENS ==========\n")
print(filtered_tokens)
```
