## Experiment 4: Term Frequency and NER


4.1. To perform Term-Frequency Analysis and Named Entity Recognition on a given text using any toolkit.

- Calculate the frequency of each word and save the output in CSV format (Term, Frequency)
- Identify the top 10 most frequent terms and display their corresponding frequencies.


```python
import csv
from collections import Counter
import nltk
import spacy

nltk.download("punkt")
nltk.download("punkt_tab")

nlp = spacy.load("en_core_web_sm")

with open("4.1_4.2_input.txt", "r", encoding="utf-8") as file:
    text = file.read()

tokens = nltk.word_tokenize(text.lower())
words = [word for word in tokens if word.isalnum()]

frequency = Counter(words)

with open("term_frequency_4_1.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["Term", "Frequency"])

    for term, freq in frequency.most_common():
        writer.writerow([term, freq])

print("Top 10 Most Frequent Terms:")
for term, freq in frequency.most_common(10):
    print(f"{term}: {freq}")

doc = nlp(text)

print("\nNamed Entities:")
for entity in doc.ents:
    print(f"{entity.text} -> {entity.label_}")
```

4.2. To perform Term-Frequency Analysis on a given text without using any toolkit.

- Calculate the frequency of each word and save the output in CSV format (Term, Frequency)
- Identify the top 10 most frequent terms and display their corresponding frequencies.


```python
import re
import csv
from collections import Counter

with open("4.1_4.2_input.txt", "r", encoding="utf-8") as file:
    text = file.read().lower()

words = re.findall(r"\b[a-zA-Z0-9]+\b", text)

frequency = Counter(words)

with open("term_frequency_4_2.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["Term", "Frequency"])

    for term, freq in frequency.most_common():
        writer.writerow([term, freq])

print("Top 10 Most Frequent Terms:")

for term, freq in frequency.most_common(10):
    print(f"{term}: {freq}")
```

4.3. Write a Python program without using NLTK, spaCy, or any other NLP toolkit to:

- Read multiple text documents.
- Convert text to lowercase.
- Remove punctuation.
- Tokenize the documents.
- Calculate TF for every term.
- Calculate DF for every term.
- Calculate IDF using the logarithmic formula.
- Calculate TF-IDF for every term in every document.
- Display Term, TF, DF, IDF, and TF-IDF at each step.
- Identify the top 10 terms with the highest TF-IDF score for each document.


```python
import re
import math
from collections import Counter

documents = {
    "Document 1": "Natural language processing is a field of artificial intelligence.",
    "Document 2": "Natural language processing helps computers understand human language.",
    "Document 3": "Machine learning is an important part of artificial intelligence."
}

tokenized = {}

for name, text in documents.items():
    text = text.lower()
    text = re.sub(r"[^\w\s]", "", text)
    tokenized[name] = text.split()

df = Counter()

for words in tokenized.values():
    df.update(set(words))

N = len(documents)

results = {}

for name, words in tokenized.items():
    tf = Counter(words)
    total_words = len(words)
    results[name] = {}

    for term, frequency in tf.items():
        term_tf = frequency / total_words
        term_df = df[term]
        term_idf = math.log(N / term_df)
        term_tfidf = term_tf * term_idf

        results[name][term] = {
            "TF": term_tf,
            "DF": term_df,
            "IDF": term_idf,
            "TF-IDF": term_tfidf
        }

print("TF-IDF RESULTS\n")

for name, terms in results.items():
    print(name)
    print("-" * 65)
    print(f"{'Term':<15}{'TF':<12}{'DF':<8}{'IDF':<12}{'TF-IDF':<12}")

    for term, values in sorted(terms.items()):
        print(
            f"{term:<15}"
            f"{values['TF']:<12.4f}"
            f"{values['DF']:<8}"
            f"{values['IDF']:<12.4f}"
            f"{values['TF-IDF']:<12.4f}"
        )

    print()

print("TOP 10 TERMS BY TF-IDF\n")

for name, terms in results.items():
    top_terms = sorted(
        terms.items(),
        key=lambda x: x[1]["TF-IDF"],
        reverse=True
    )[:10]

    print(name)

    for term, values in top_terms:
        print(f"{term}: {values['TF-IDF']:.4f}")

    print()
```
