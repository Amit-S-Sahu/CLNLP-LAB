5.1. Implement subword-level tokenization using a BPE-based tokenizer.

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

file_path = "/content/Experiment-5/input_sub_word_data.txt"
with open(file_path, "r", encoding="utf-8") as f:
    text = f.read().strip()

tokens = tokenizer.tokenize(text)
token_ids = tokenizer.convert_tokens_to_ids(tokens)

print("Tokens:", tokens)
print("Token IDs:", token_ids)

for token, token_id in zip(tokens, token_ids):
    print(f"{token:15} {token_id}")
```

```python
from tokenizers import Tokenizer, models, trainers, pre_tokenizers

file_path = "/content/Experiment-5/input_sub_word_data.txt"
with open(file_path, "r", encoding="utf-8") as f:
    text = f.read().strip()

tokenizer = Tokenizer(models.BPE(unk_token="[UNK]"))
tokenizer.pre_tokenizer = pre_tokenizers.Whitespace()

trainer = trainers.BpeTrainer(
    vocab_size=1000,
    min_frequency=2,
    special_tokens=["[UNK]"]
)

tokenizer.train_from_iterator([text], trainer)

sentence = text.splitlines()[0]
output = tokenizer.encode(sentence)

print("Sentence:", sentence)
print("Tokens:", output.tokens)
print("Token IDs:", output.ids)

for token, token_id in zip(output.tokens, output.ids):
    print(f"{token:15} {token_id}")
```

 5.2. Implement subword-level tokenization using a SentencePiece tokenizer.

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("t5-small")

file_path = "/content/Experiment-5/input_sub_word_data.txt"
with open(file_path, "r", encoding="utf-8") as f:
    text = f.read().strip()

tokens = tokenizer.tokenize(text)
token_ids = tokenizer.convert_tokens_to_ids(tokens)

print("Tokens:", tokens[:30], "... (truncated)" if len(tokens) > 30 else "")
print("Token IDs:", token_ids[:30], "... (truncated)" if len(token_ids) > 30 else "")

for token, token_id in zip(tokens[:20], token_ids[:20]):
    print(f"{token:15} {token_id}")
```

```python
import sentencepiece as spm

input_file = "/content/Experiment-5/input_sub_word_data.txt"

spm.SentencePieceTrainer.train(
    input=input_file,
    model_prefix="my_tokenizer",
    vocab_size=200,
    model_type="bpe",
    hard_vocab_limit=False
)

tokenizer = spm.SentencePieceProcessor(
    model_file="my_tokenizer.model"
)

with open(input_file, "r", encoding="utf-8") as f:
    text = f.read()

sentence = next(line for line in text.splitlines() if line.strip())

tokens = tokenizer.encode(sentence, out_type=str)
token_ids = tokenizer.encode(sentence, out_type=int)

print("Sentence:", sentence)
print("Tokens:", tokens)
print("Token IDs:", token_ids)

for token, token_id in zip(tokens, token_ids):
    print(f"{token:15} {token_id}")
```

  5.3. Implement Part-of-Speech (POS) tagging on a given text and display the unique token, POS tag, and description of each POS tag.
                            
                   "The young student is reading an interesting book in the library."   
 (Implement two separate codes: a. Using Spacy and b.  NLTK )


```python
import spacy

nlp = spacy.load("en_core_web_sm")
text = "The young student is reading an interesting book in the library."

doc = nlp(text)

print(f"{'Token':15} {'POS Tag':12} Description")
print("-" * 55)

for token in doc:
    if not token.is_space:
        print(f"{token.text:15} {token.pos_:12} {spacy.explain(token.pos_)}")
```

```python
import nltk

nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger_eng")
nltk.download("universal_tagset")

text = "The young student is reading an interesting book in the library."

tokens = nltk.word_tokenize(text)
tagged = nltk.pos_tag(tokens)

descriptions = {
    "DT": "Determiner",
    "JJ": "Adjective",
    "NN": "Noun, singular",
    "NNS": "Noun, plural",
    "VB": "Verb, base form",
    "VBD": "Verb, past tense",
    "VBG": "Verb, gerund or present participle",
    "VBN": "Verb, past participle",
    "VBP": "Verb, non-3rd-person singular present",
    "VBZ": "Verb, 3rd-person singular present",
    "IN": "Preposition or subordinating conjunction",
    "RB": "Adverb",
    "PRP": "Personal pronoun",
    ".": "Punctuation"
}

print(f"{'Token':15} {'POS Tag':12} Description")
print("-" * 55)

for token, tag in tagged:
    print(f"{token:15} {tag:12} {descriptions.get(tag, 'Other')}")
```

5.4. Implement Part-of-Speech (POS) tagging on a given text and display the unique token, POS tag, description of each token and frequency using Spacy.


```python
import spacy
from collections import Counter

nlp = spacy.load("en_core_web_sm")

file_path = "/content/Experiment-5/input_sub_word_data.txt"
with open(file_path, "r", encoding="utf-8") as f:
    text = f.read()

doc = nlp(text)

counts = Counter(
    (token.text.lower(), token.pos_)
    for token in doc
    if not token.is_space and not token.is_punct
)

print(f"{'Token':15} {'POS Tag':12} {'Description':30} Frequency")
print("-" * 75)

for (word, pos), frequency in sorted(counts.items()):
    print(
        f"{word:15} {pos:12} "
        f"{spacy.explain(pos) or 'Unknown':30} {frequency}"
    )
```

5.7 Train a Word2Vec model using the provided corpus and calculate the cosine similarity between selected word pairs.

Calculate similarity for:
- machine and learning
- natural and language
- deep and learning


```python
from gensim.models import Word2Vec

corpus_path = "corpus.txt"
sentences = []
with open(corpus_path, "r", encoding="utf-8") as f:
    for line in f:
        if line.strip():
            sentences.append(line.lower().split())

model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=1,
    sg=0,
    workers=1,
    seed=42,
    epochs=100
)

pairs = [
    ("machine", "learning"),
    ("natural", "language"),
    ("deep", "learning")
]

print(f"{'Word Pair':30} Cosine Similarity")
print("-" * 50)

for word1, word2 in pairs:
    similarity = model.wv.similarity(word1, word2)
    print(f"{word1 + ' - ' + word2:30} {similarity:.4f}")
```
