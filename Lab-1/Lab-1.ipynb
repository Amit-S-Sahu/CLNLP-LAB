### Experiment-1: Data Analysis

Experiment-1.1: Read the employee dataset and display the average salary of each department using a bar chart.

Experiment 1.2: Display the number of employees working in each department.

Experiment 1.3:Display the percentage of male and female employees using a pie chart

Experiment 1.4: Draw a histogram showing salary distribution.

Experiment 1.5: Draw a scatter plot showing the relationship between experience and salary.

Experiment 1.6: Display the top 10 employees with the highest salary.

Experiment 1.7: Find the highest salary in every department.

Experiment 1.8: Display employees whose salary is greater than the overall average salary.

Experiment 1.9: Calculate the average years of experience in each department.

Experiment 1.10: Display the distribution of employee ages using a histogram.


```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("employee_information_100.csv")

df["Age"] = pd.to_numeric(df["Age"])
df["Salary"] = pd.to_numeric(df["Salary"])
df["Experience_Years"] = pd.to_numeric(df["Experience_Years"])

print("Employee Dataset:")
print(df)
```

```python
avg_salary = df.groupby("Department")["Salary"].mean()

print("\n1.1 Average Salary of Each Department:")
print(avg_salary)

avg_salary.plot(kind="bar")

plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
employee_count = df["Department"].value_counts()

print("\n1.2 Number of Employees in Each Department:")
print(employee_count)
```

```python
gender_percentage = df["Gender"].value_counts()

print("\n1.3 Percentage of Male and Female Employees:")
print((gender_percentage / len(df) * 100).round(2))

plt.pie(
    gender_percentage,
    labels=gender_percentage.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Percentage of Male and Female Employees")
plt.show()
```

```python
plt.hist(df["Salary"], bins=10, edgecolor="black")

plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")
plt.tight_layout()
plt.show()
```

```python
plt.scatter(df["Experience_Years"], df["Salary"])

plt.title("Relationship Between Experience and Salary")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")
plt.tight_layout()
plt.show()
```

```python
top_10 = df.nlargest(10, "Salary")

print("\n1.6 Top 10 Employees with Highest Salary:")
print(top_10[["Employee_ID", "Name", "Department", "Salary"]])
```

```python
highest_salary = df.groupby("Department")["Salary"].max()

print("\n1.7 Highest Salary in Every Department:")
print(highest_salary)
```

```python
overall_average_salary = df["Salary"].mean()

employees_above_average = df[df["Salary"] > overall_average_salary]

print("\n1.8 Overall Average Salary:")
print(overall_average_salary)

print("\nEmployees with Salary Greater Than Overall Average:")
print(
    employees_above_average[
        ["Employee_ID", "Name", "Department", "Salary"]
    ]
)
```

```python
avg_experience = df.groupby("Department")["Experience_Years"].mean()

print("\n1.9 Average Years of Experience in Each Department:")
print(avg_experience)
```

```python
plt.hist(df["Age"], bins=10, edgecolor="black")

plt.title("Distribution of Employee Ages")
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.tight_layout()
plt.show()
```
