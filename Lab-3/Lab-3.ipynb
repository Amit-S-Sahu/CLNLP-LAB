### Experiment-3: Stemming, Lemmatization and Regular Expression


Experiment 3.1: Write a Python program to perform Write a Python program to perform stemming using the Porter Stemmer. Apply stemming to the following words. Display the original words and their stems.

playing, played, plays, studies, studying, connected, connection, computers


```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()

words = [
    "playing", "played", "plays", "studies",
    "studying", "connected", "connection", "computers"
]

print("Original Word\tStem")
print("-------------------------")

for word in words:
    stem = ps.stem(word)
    print(f"{word}\t\t{stem}")
```

Experiment 3.2: Write a Python program to perform lemmatization on a given input words:

cats, dogs, running, runs, ran, studies, studying, better, children, mice, went, ate,
leaves, caring,


```python
import nltk
from nltk.stem import WordNetLemmatizer

nltk.download("wordnet")
nltk.download("omw-1.4")

lemmatizer = WordNetLemmatizer()

words = [
    "cats", "dogs", "running", "runs", "ran",
    "studies", "studying", "better", "children",
    "mice", "went", "ate", "leaves", "caring"
]

pos_tags = [
    "n", "n", "v", "v", "v",
    "v", "v", "a", "n",
    "n", "v", "v", "n", "v"
]

print("Original Word\tLemma")
print("-------------------------")

for word, pos in zip(words, pos_tags):
    lemma = lemmatizer.lemmatize(word, pos=pos)
    print(f"{word}\t\t{lemma}")
```

Experiment 3.2: Write a Python program using Regular Expressions (Regex) to extract and process the following information from a text:

- Email addresses
- URLs
- Mobile numbers
- Hashtags
- Mentions


```python
import re

text = """
Welcome to the Natural Language Processing workshop! For more information,
visit https://www.nlpworkshop.com or www.python.org. You can contact the
coordinator at nlpworkshop@gmail.com or support@python.org. For registration,
call +91-9876543210 or 9123456789. Follow us on social media @NLPWorkshop
and @PythonLearner. Share your experience using #NLP, #Python, and
#MachineLearning. You can also visit https://github.com/NLPWorkshop
for the latest updates.
"""

emails = re.findall(
    r'[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}',
    text
)

urls = re.findall(
    r'(?:https?://|www\.)[A-Za-z0-9.-]+\.[A-Za-z]{2,}(?:/[A-Za-z0-9_./-]*)?',
    text
)

mobile_numbers = re.findall(
    r'(?:\+91[-\s]?)?[6-9]\d{9}',
    text
)

hashtags = re.findall(
    r'#[A-Za-z0-9_]+',
    text
)

mentions = re.findall(
    r'@[A-Za-z0-9_]+',
    text
)

print("Email Addresses:")
for email in emails:
    print(email)

print("\nURLs:")
for url in urls:
    print(url)

print("\nMobile Numbers:")
for number in mobile_numbers:
    print(number)

print("\nHashtags:")
for hashtag in hashtags:
    print(hashtag)

print("\nMentions:")
for mention in mentions:
    print(mention)
```
